```python
skor_nobita = 0

def update_skor(nilai):
    global skor_nobita
    skor_nobita += nilai


def belajar(jam_belajar):
    if jam_belajar < 1:
        print("Nobita hanya membuka buku sebentar")
        return 50
    elif jam_belajar < 3:
        print("Nobita belajar secukupnya")
        return 25
    else:
        print("Nobita belajar dengan rajin hari ini")
        return 10


def tidur(jam_tidur):
    if jam_tidur < 1:
        print("Nobita kurang tidur")
        return -10
    elif jam_tidur <= 2:
        print("Waktu tidur Nobita seimbang")
        return 20
    else:
        print("Nobita tidur terlalu lama")
        return -20


def main_bersama_doraemon(jumlah_gadget, waktu_main):
    if waktu_main <= 1 and jumlah_gadget <= 2:
        print("Nobita bermain secukupnya")
        return 15
    elif waktu_main <= 2 and jumlah_gadget <= 3:
        print("Nobita masih bisa mengatur waktu dengan baik")
        return 5
    else:
        print("Nobita kebanyakan bermain")
        return -30


def menyontek():
    print("Nobita ketahuan menyontek")
    return -50


def jalankan_aktivitas(pilihan):
    if pilihan == 1:
        jam_belajar = float(input("Berapa jam Nobita belajar hari ini? "))
        nilai = belajar(jam_belajar)
        update_skor(nilai)

    elif pilihan == 2:
        jam_tidur = float(input("Berapa jam Nobita tidur siang hari ini? "))
        nilai = tidur(jam_tidur)
        update_skor(nilai)

    elif pilihan == 3:
        jumlah_gadget = int(input("Berapa gadget yang digunakan Nobita? "))
        waktu_main = float(input("Berapa jam Nobita bermain gadget? "))
        nilai = main_bersama_doraemon(jumlah_gadget, waktu_main)
        update_skor(nilai)

    elif pilihan == 4:
        nilai = menyontek()
        update_skor(nilai)

    else:
        print("Pilihan tidak valid, aktivitas hanya ada 4 pilihan")


def evaluasi_akhir():
    if skor_nobita >= 80:
        return "Siswa Teladan"
    elif skor_nobita >= 50:
        return "Cukup Rajin"
    else:
        return "Pemalas"


# F. Menu Utama Program
def menu_utama():
    while True:
        print("\n===================================")
        print(" PROGRAM PENINGKAT DISIPLIN HARIAN")
        print("===================================")
        print("1. Belajar")
        print("2. Tidur Siang")
        print("3. Bermain Bersama Doraemon")
        print("4. Menyontek")
        print("5. Lihat Skor dan Evaluasi")
        print("6. Keluar")
        print("===================================")

        pilihan = int(input("Pilih aktivitas Nobita (1-6): "))

        if pilihan >= 1 and pilihan <= 4:
            jalankan_aktivitas(pilihan)
            print("Skor Nobita saat ini:", skor_nobita)

        elif pilihan == 5:
            print("\n========== EVALUASI AKHIR ==========")
            print("Total skor Nobita:", skor_nobita)
            print("Hasil evaluasi:", evaluasi_akhir())

        elif pilihan == 6:
            print("\nProgram selesai.")
            print("Total skor Nobita:", skor_nobita)
            print("Evaluasi akhir:", evaluasi_akhir())
            break

        else:
            print("Pilihan tidak valid!")


# Menjalankan Program
menu_utama()
```
