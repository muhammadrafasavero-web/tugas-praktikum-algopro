```python
import math


# Data diri Giant
nama = "Giant"
usia = 20
berat_badan = 97.5
tinggi_badan = 1.75
status_olahraga = False

# Konversi berat dari string ke float
berat_catatan = float("95.7")

print(
    f"Halo, saya {nama}. Usia saya {usia} tahun "
    f"dan berat saya {berat_catatan} kg"
)

# Menghitung BMI
bmi = berat_badan / tinggi_badan**2

print(f"BMI asli: {bmi}")
print(f"BMI dibulatkan: {bmi:.2f}")

# Target minum air
target_air = 2000
ukuran_botol = 350

jumlah_botol_desimal = target_air / ukuran_botol
jumlah_botol_penuh = target_air // ukuran_botol
sisa_air = target_air % ukuran_botol

print(f"Jumlah botol secara true division: {jumlah_botol_desimal}")
print(f"Jumlah botol penuh: {jumlah_botol_penuh}")
print(f"Sisa air: {sisa_air} ml")

# Menghitung kebutuhan kalori
kalori_harian = 2100.5

kalori_bawah = math.floor(kalori_harian)
kalori_atas = math.ceil(kalori_harian)

print(f"Kalori dibulatkan ke bawah: {kalori_bawah}")
print(f"Kalori dibulatkan ke atas: {kalori_atas}")
```
