```python
def compound_interest(principal, rate, years):
    return principal * (1 + rate) ** years


def simple_interest(principal, rate, years):
    return principal + (principal * rate * years)


def print_result(principal, rate, years, ci, si):
    print("\n===== HASIL PERHITUNGAN TABUNGAN =====")
    print(f"Saldo Awal              : ¥{principal:,.2f}")
    print(f"Bunga Tahunan           : {rate * 100}%")
    print(f"Lama Menabung           : {years} tahun")
    print(f"Nilai Akhir (Majemuk)   : ¥{ci:,.2f}")
    print(f"Nilai Akhir (Sederhana) : ¥{si:,.2f}")


def main():
    print("==============================")
    print("PROGRAM KALKULATOR BUNGA TABUNGAN")
    print("==============================")

    principal = float(input("Masukkan saldo awal: "))
    rate = float(input("Masukkan bunga tahunan (%): ")) / 100
    years = int(input("Masukkan lama menabung (tahun): "))

    # Panggil fungsi lokal secara langsung
    ci = compound_interest(principal, rate, years)
    si = simple_interest(principal, rate, years)

    # Panggil fungsi cetak lokal secara langsung
    print_result(principal, rate, years, ci, si)


if __name__ == "__main__":
    main()
```
