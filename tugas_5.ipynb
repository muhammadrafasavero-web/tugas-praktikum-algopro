```python
hari = 1
saldo = 50000

while hari <= 30:
    pengeluaran = int(input(f"Masukkan pengeluaran hari ke-{hari}: "))
    saldo -= pengeluaran

    if saldo < 0:
        break

    print(f"Sisa saldo Nobita: Rp{saldo}")
    hari += 1
else:
    print("Pencatatan selama 30 hari selesai.")

if saldo < 0:
    print("Saldo Nobita habis! Pencatatan dihentikan.")
```

```python
pengeluaran_mingguan = [20000, 0, 15000, 0]

for i in range(4):
    if pengeluaran_mingguan[i] == 0:
        continue

    print(f"Pengeluaran minggu ke-{i + 1}: Rp{pengeluaran_mingguan[i]}")
else:
    print("Analisis pengeluaran mingguan selesai.")
```

```python
for bulan in range(1, 4):
    total = 0

    print(f"\nBulan ke-{bulan}")

    for minggu in range(1, 5):
        pengeluaran = int(input(f"Masukkan pengeluaran minggu ke-{minggu}: "))

        if pengeluaran > 20000:
            print("Pengeluaran melebihi Rp20.000. Bulan dihentikan.")
            break

        total += pengeluaran

    else:
        print(f"Total pengeluaran bulan ke-{bulan}: Rp{total}")

print("\nLaporan bulanan selesai.")
```

```python
saldo_awal = 300000
saldo = saldo_awal

for bulan in range(11, 12):
    minggu = 1

    print(f"\nBulan ke-{bulan}")

    while minggu <= 4:
        libur = input(f"Apakah minggu ke-{minggu} libur? (y/n): ").lower()

        if libur == "y":
            minggu += 1
            continue

        elif libur == "n":
            pengeluaran = int(input("Masukkan pengeluaran: Rp"))
            saldo -= pengeluaran

            if saldo < 0:
                print("Saldo negatif! Simulasi dihentikan.")
                break

            minggu += 1

        else:
            print("Input tidak valid. Masukkan y atau n.")
            continue

    else:
        print(f"Saldo akhir bulan ke-{bulan}: Rp{saldo}")

    if saldo < 0:
        break

else:
    print(f"\nNobita berhasil menabung {saldo} dari total {saldo_awal}")
```
