```python
usia_pengemudi = int(input("Masukkan usia pengemudi: "))
jenis_kendaraan = input("Jenis kendaraan (mobil/motor): ").lower()
nilai_kendaraan = float(input("Nilai kendaraan (Rp): "))
area_penggunaan = input("Area penggunaan (perkotaan/pedesaan): ").lower()
sistem_pengaman = input("Memiliki sistem pengaman? (ya/tidak): ").lower()

if usia_pengemudi < 20 or usia_pengemudi > 60:
    risiko_usia = "tinggi"
else:
    risiko_usia = "rendah"

if jenis_kendaraan == "motor":
    risiko_kendaraan = "sedang"
else:
    risiko_kendaraan = "rendah"

if risiko_usia == "tinggi" or risiko_kendaraan == "tinggi":
    risiko = "tinggi"
elif risiko_usia == "sedang" or risiko_kendaraan == "sedang":
    risiko = "sedang"
else:
    risiko = "rendah"

tingkat_risiko = ["rendah", "sedang", "tinggi"]

if area_penggunaan == "perkotaan" and risiko != "tinggi":
    risiko = tingkat_risiko[tingkat_risiko.index(risiko) + 1]

if sistem_pengaman == "tidak":
    if risiko == "rendah":
        risiko = "sedang"
    elif risiko == "sedang":
        risiko = "tinggi"

if risiko == "tinggi":
    persentase_premi = 0.10
elif risiko == "sedang":
    persentase_premi = 0.07
else:
    persentase_premi = 0.05

premi = persentase_premi * nilai_kendaraan

pesan = ("Doraemon menyarankan asuransi tambahan untuk perlindungan penuh!"
    if risiko == "tinggi"
    else "Risiko kendaraan masih aman menurut Doraemon")

print("\n=== Smart Vehicle Insurance Analyzer ===")
print("Kategori Risiko :", risiko)
print(f"Premi Asuransi  : Rp{premi:,.0f}")
print("Pesan Doraemon  :", pesan)
```
